```python
import pandas as pd 
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('Titanic.csv') 
```

```python
df
```

```python
# Drop irrelevant columns
df = df[["Survived", "Pclass", "Sex", "Age", "Embarked"]]

# Drop missing values
df = df.dropna()

# Encode categorical variable
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
df["Embarked"] = df["Embarked"].map({"C":0, "Q":1, "S":2})
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X = df[["Pclass", "Sex", "Age","Embarked"]]
y = df["Survived"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()
model.fit(X_train, y_train)
# model.fit(X,y)
```

<h1> Accuracy Test </h1>

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

<h1> TITANIC SURVAVILITY USING LOGISTIC REGRESSION </h1>

```python
df2 = pd.read_csv("Titanic.csv")
df2
```

```python
# Keep only useful columns
df2 = df2[["Survived", "Pclass", "Sex", "Age","Fare" ,"Embarked"]]

# Drop missing values
df2 = df2.dropna()
```

```python
df2
```

```python
# Encode 'Sex' and 'Embarked'
df2["Sex"] = df2["Sex"].map({"male": 0, "female": 1})
df2["Embarked"] = df2["Embarked"].map({"S": 0, "C": 1, "Q": 2})
```

```python
df2
```

<h1>selecting X and y </h1>

```python
X = df2[["Pclass", "Sex", "Age", "Fare", "Embarked"]]
y = df2["Survived"]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

<h1> Manual Testing </h1>

```python
# [Pclass=3, Sex=0 (male), Age=22]
model.predict([[1, 0, 1, 0]])
```

```python
p1 = model.predict([[3, 0, 5, 7.25]])   
p2 = model.predict([[2, 0,43, 14.0]])   

if p1 == [1]:
    print("Passenger likely survive")
elif p2 == [0]:
    print("Passenger unlikely to survive")
else:
    print("Invalid Result")
```

<h2>Insight: The model demonstrates that the strong positive weight given to children (low Age) is enough to overcome the negative weights from $\mathbf{Sex}$ (Male) and $\mathbf{Pclass}$ (3rd), illustrating how the Log-Odds calculation in Logistic Regression reflects the historic survival priorities ("women and children first").</h2>
