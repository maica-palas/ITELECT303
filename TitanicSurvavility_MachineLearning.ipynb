```python
import pandas as pd 
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('Titanic.csv') 
```

```python
df
```

```python
# Drop irrelevant columns
df = df[["Survived", "Pclass", "Sex", "Age", "Embarked"]]

# Drop missing values
df = df.dropna()

# Encode categorical variable
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
df["Embarked"] = df["Embarked"].map({"C":0, "Q":1, "S":2})
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X = df[["Pclass", "Sex", "Age","Embarked"]]
y = df["Survived"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()
model.fit(X_train, y_train)
# model.fit(X,y)
```

<h1> Accuracy Test </h1>

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

<h1> TITANIC SURVAVILITY USING LOGISTIC REGRESSION </h1>

```python
df2 = pd.read_csv("Titanic.csv")
df2
```

```python
# Keep only useful columns
df2 = df2[["Survived", "Pclass", "Sex", "Age","Fare" ,"Embarked"]]

# Drop missing values
df2 = df2.dropna()
```

```python
df2
```

```python
# Encode 'Sex' and 'Embarked'
df2["Sex"] = df2["Sex"].map({"male": 0, "female": 1})
df2["Embarked"] = df2["Embarked"].map({"S": 0, "C": 1, "Q": 2})
```

```python
df2
```

<h1>selecting X and y </h1>

```python
X = df2[["Pclass", "Sex", "Age", "Fare", "Embarked"]]
y = df2["Survived"]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

<h1> Manual Testing </h1>

```python
AGE = 22.0
EMBARKED = 0

passengers = {
    "A: Wealthy Man (1st Class, Male, Age 22, High Fare)": [1, 0, 22, 150.0, 0],
    "B: Wealthy Woman (1st Class, Female, Age 22, High Fare)": [1, 1, 22, 150.0, 0],
    "C: Poor Man (3rd Class, Male, Age 22, Low Fare)": [3, 0, 22, 7.50, 0],
    "D: Poor Woman (3rd Class, Female, Age 22, Low Fare)": [3, 1, 22, 7.50, 0]
}

print("\n--- Gender and Wealth Prediction Comparison (Age 22) ---")
for desc, features in passengers.items():
    prediction = model.predict([features])[0]
    if prediction == 1:
        print(f"{desc} → Likely to survive ")
    else:
        print(f"{desc} → Unlikely to survive ")
```

<h3> Insight A : The model makes an exception to save this man. Even though being male is a huge risk, his very high ticket price acts like a pass that moves him to the front of the line.</h3>

<h3>Insight B: This passenger has the best possible chance. She is a woman (the biggest advantage) and rich (the second biggest advantage). The model is completely certain she survives.</h3>

<h3>Insight C: This is the lowest priority passenger. He is a man and has the cheapest ticket. The model sees nothing positive to overcome the double disadvantage, so it predicts he can't survive.</h3>

<h3> Insight D: The model finds that being a woman is so powerful that it does not care about her 3rd Class ticket or low fare. Gender guarantees her survival prediction.</h3>
